# Single-sample debugging
Run from the repository root. Change `CONFIG` to test Qwen or an API provider.


```python
from pathlib import Path
from IPython.display import display
from PIL import Image
from visionbench.config import load_config
from visionbench.plugins import load_dataset, load_evaluator
from visionbench.runner import infer
```

```python
CONFIG = "config_dummy"
config = load_config(CONFIG)
entry = config.data["dataset"]
dataset = load_dataset(config.scoped_path("dataset", entry["path"]), {**entry.get("params", {}), "raw_dir": str(config.scoped_path("raw", entry.get("raw_path", "."))), "assets_dir": str(config.scoped_path("assets", entry.get("assets_path", ".")))})
sample = dataset[0]
print(sample.sample_id, sample.prompt, sample.image.size)
display(sample.image)
```

```python
output_path = infer(config, config.scoped_path("raw", "white_background/sample_01.png"), sample.prompt)
output = Image.open(output_path)
display(output)
print(output_path)
```

```python
entry = config.data["evaluator"]
evaluator = load_evaluator(config.scoped_path("evaluator", entry["path"]), entry.get("params", {}))
print(evaluator.score(evaluator.prepare_output(output, sample), sample))
```
